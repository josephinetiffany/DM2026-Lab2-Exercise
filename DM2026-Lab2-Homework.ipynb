# **Competition Code**

## 1. Preprocessing Steps

```python
### Add the code related to the preprocessing steps in cells inside this section
import json
import pandas as pd

DATA_DIR = "data/kaggle"  # Kaggle files (git-ignored; download them from the competition page)

# Load the posts, the train/test split, and the training labels.
with open(f"{DATA_DIR}/final_posts.json", encoding="utf-8") as f:
    posts = pd.DataFrame(json.load(f))                     # columns: post_id, text
split = pd.read_csv(f"{DATA_DIR}/data_identification.csv")   # columns: post_id, set
labels = pd.read_csv(f"{DATA_DIR}/emotion.csv")              # columns: post_id, label

# Attach the split to every post, then attach labels to the training posts.
df = posts.merge(split, on="post_id")
train_df = df[df["set"] == "train"].merge(labels, on="post_id").reset_index(drop=True)
test_df = df[df["set"] == "test"].reset_index(drop=True)

# Light cleaning: drop missing text and trim whitespace. Case and punctuation are kept
# here because the vectorizers below handle lowercasing, and "!!" or "?!" carry emotion.
for part in (train_df, test_df):
    part["text"] = part["text"].fillna("").str.strip()

print("train:", train_df.shape, " test:", test_df.shape)
print(train_df["label"].value_counts())
```

## 2. Feature Engineering Steps

```python
### Add the code related to the feature engineering steps in cells inside this section
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import make_union

# Two TF-IDF views of each post, concatenated side by side:
#  - word unigrams + bigrams catch emotion words and short phrases ("so happy", "not fair")
#  - character 2-5 grams (within word boundaries) cope with typos, slang and elongated
#    words ("sooo", "hahaha") that word features would treat as unseen tokens.
# sublinear_tf dampens words repeated many times in one post.
features = make_union(
    TfidfVectorizer(ngram_range=(1, 2), sublinear_tf=True),
    TfidfVectorizer(analyzer="char_wb", ngram_range=(2, 5), min_df=2, sublinear_tf=True),
)
```

## 3. Model Implementation Steps

```python
### Add the code related to the model implementation steps in cells inside this section
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.pipeline import make_pipeline
from sklearn.svm import LinearSVC

# Linear SVM on the TF-IDF features. class_weight="balanced" up-weights rare classes
# (fear has about a third as many posts as joy), which helps macro-F1.
# C=0.2 was the best of 0.05 / 0.1 / 0.2 / 0.5 in 5-fold cross-validation.
clf = make_pipeline(features, LinearSVC(C=0.2, class_weight="balanced"))

# Estimate leaderboard performance with stratified 5-fold CV on the training set.
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scores = cross_val_score(clf, train_df["text"], train_df["label"], cv=cv, scoring="f1_macro")
print(f"5-fold macro-F1: {scores.mean():.4f} +/- {scores.std():.4f}")
```

```python
# Refit on all training posts, predict the test posts, and write the Kaggle file.
clf.fit(train_df["text"], train_df["label"])
submission = pd.DataFrame({"post_id": test_df["post_id"], "label": clf.predict(test_df["text"])})

# Keep the row order of the sample submission so the file matches Kaggle's format exactly.
sample = pd.read_csv(f"{DATA_DIR}/sampleSubmission.csv")
submission = sample[["post_id"]].merge(submission, on="post_id", how="left")
assert submission["label"].notna().all() and len(submission) == len(sample)

# Kaggle expects the ID column to be named "id" (the sample file says post_id, but the grader rejects it).
submission = submission.rename(columns={"post_id": "id"})
submission.to_csv("submission.csv", index=False)
print(submission["label"].value_counts())
```
